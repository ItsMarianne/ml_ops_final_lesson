<font color='black'><div class="alert alert-info"><strong>**5. Train / test split**</strong><br>

We hold out **20%** of data the model **never sees** during training. Final scores on this set estimate real-world performance.

**Rule:** Features `X` and target `y` are split together so each row stays intact.

````python
X = df[FEATURE_NAMES]
y = df[TARGET]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=RANDOM_STATE) # unpacking

print(f"Training rows: {len(X_train):,}")
print(f"Test rows:     {len(X_test):,}")
````

<font color='black'><div class="alert alert-info"><strong>**6. Feature scaling (RobustScaler)**</strong><br>

**Why do we scale?** Ridge regression treats large-magnitude features as more important. `Population` (thousands) vs `HouseAge` (~30) are on different scales.

**RobustScaler** subtracts the median and divides by the inter-quartile range — it handles outliers better than StandardScaler (important in finance-style data).

We wrap scaling + model in a **Pipeline** so the same steps run at training and prediction time — no train/serve skew.

<font color='black'><div class="alert alert-warning"><strong>**Concept: what is a `Pipeline`?**</strong><br>

```python
# Without a Pipeline
scaler = RobustScaler() # Instantiate the RobustScaler class
model = Ridge(alpha=1.0)

scaled_train_data = scaler.fit_transform(X_train)
scaled_test_data = scaler.transform(X_test) # preventing data leakage

model.fit(scaled_train_data, y_train)
model.predict(scaled_test_data)
```

A **Pipeline** chains several steps into a **single object that behaves like one model**. Every step except the last *transforms* the data; the last step is the *estimator* (the model itself).

```python
# With a Pipeline
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import RobustScaler
from sklearn.linear_model import Ridge

pipe = Pipeline([
    ("scaler", RobustScaler()),    # step 1: scale the features
    ("model",  Ridge(alpha=1.0)),  # step 2: fit the model on the scaled features
])
pipe.fit(X_train, y_train)         # runs scaler.fit_transform → model.fit, in order
predictions = pipe.predict(X_test) # runs scaler.transform → model.predict, automatically
```

**Why use it instead of scaling by hand?**

| Problem without a Pipeline | How the Pipeline fixes it |
|---|---|
| Easy to leak test data into the scaler | The scaler learns its numbers (median/IQR) from **training data only**, then reuses them everywhere |
| Train/serve skew — different prep in prod | The **same steps** run in training and production |
| Messy code, easy to forget a step | One `.fit()` / `.predict()` call does everything |

when we move to prod in later session:
> You save and load the *whole pipeline* (`joblib.dump(pipe, ...)`). This guarantees live data is always prepared the exact same way it was during training — no "I forgot to scale the input" mistake that is very common.

````python
def evaluate_regression(y_true, y_pred, label="Model"):
    """Return dict of standard regression metrics."""
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mae = mean_absolute_error(y_true, y_pred)
    r2 = r2_score(y_true, y_pred)
    print(f"\n=== {label} ===")
    print(f"  RMSE : {rmse:.4f}  (typical error in $100k units → multiply by 100k for USD)")
    print(f"  MAE  : {mae:.4f}  (average absolute error)")
    print(f"  R²   : {r2:.4f}  (1.0 = perfect; 0 = predict the mean)")
    return {"model": label, "RMSE": rmse, "MAE": mae, "R2": r2}


def plot_predictions(y_true, y_pred, title):
    plt.figure(figsize=(6, 6))
    plt.scatter(y_true, y_pred, alpha=0.3, s=10)
    lims = [min(y_true.min(), y_pred.min()), max(y_true.max(), y_pred.max())]
    plt.plot(lims, lims, "r--", lw=2, label="Perfect prediction")
    plt.xlabel("Actual")
    plt.ylabel("Predicted")
    plt.title(title)
    plt.legend()
    plt.tight_layout()
    plt.show()
````

<font color='black'><div class="alert alert-info"><strong>**7. Ridge Regression**</strong><br>

Ridge is linear regression with a penalty,is a regularization method applied in machine learning to prevent models from overfitting by applying penalties to large model coefficients which shrinks them to near zero.<br>
| Pros | Cons |
|------|------|
| Very fast to train | Cannot capture curved relationships by itself |
| Coefficients are interpretable ("+$0.05 per $1k income") | Weak if patterns are highly non-linear |
| Good baseline | |

**When to use:** First model in most ML projects, especially in finance, where regulators seek interpretability.

**Key hyperparameter:** `alpha` — higher alpha = stronger penalty = simpler model.

<font color='black'><div class="alert alert-success"><strong>**The Ridge Model Pipeline**</strong><br>

````python
ridge_pipe = Pipeline([
    ("scaler", RobustScaler()),
    ("model", Ridge(random_state=RANDOM_STATE)),
])

ridge_pipe.fit(X_train, y_train)
y_pred_ridge = ridge_pipe.predict(X_test)
````

<font color='black'><div class="alert alert-success"><strong>**Model Evaluation:- Ridge**</strong><br>

````python
ridge_metrics = evaluate_regression(y_test, y_pred_ridge, "Ridge Regression")
plot_predictions(y_test, y_pred_ridge, "Ridge: Actual vs Predicted (test set)")
````

<font color='black'><div class="alert alert-success"><strong>**Interpretability/ Feature Importance — Ridge coefficients (after scaling inside pipeline)**</strong><br>

````python
ridge_model = ridge_pipe.named_steps["model"]
coefs = pd.Series(ridge_model.coef_, index=FEATURE_NAMES).sort_values(key=abs, ascending=False)

plt.figure(figsize=(8, 4))
coefs.plot(kind="barh", color="steelblue")
plt.title("Ridge coefficients (standardized features)")
plt.xlabel("Weight")
plt.tight_layout()
plt.show()

print("Top drivers (absolute coefficient):")
print(coefs.head(5).to_string())
````

<font color='black'><div class="alert alert-warning"><strong>**Concept: what is `GridSearchCV` (hyperparameter tuning)?**</strong><br>

Models have **hyperparameters** — settings *you* choose **before** training (e.g. Ridge's `alpha`, a forest's `max_depth`). The best value is not known up front, so instead of guessing we **try several options and keep the winner that is the set of parameters that gives a better or improved evaluation result**.

- **Grid search** = try every combination of the values you list.
- **CV** = *cross-validation*, so each combination is scored on several train/validation folds — not one lucky split.

```python
# How GridSearchCV is defined/instantiated and subsequently used
from sklearn.model_selection import GridSearchCV

search = GridSearchCV(
    pipe,                                            # the Pipeline (or bare model) to tune
    param_grid={"model__alpha": [0.1, 1.0, 10.0]},   # values to try ("model__" targets the step named "model")
    scoring="neg_root_mean_squared_error",           # how "best" is measured (higher = better, so RMSE is negated)
    cv=5,                                            # 5-fold cross-validation
    n_jobs=-1,                                       # use all CPU cores
)
search.fit(X_train, y_train)

print(search.best_params_)            # the winning settings
best_model = search.best_estimator_   # pipeline refit on ALL training data using those settings
```

**How it works:** With 3 alpha values and cv=5, it trains 3 × 5 = 15 models and averages the 5 fold-scores per alpha. It then takes the best-scoring alpha and refits a single model on the full training set — ready to use.

> **Avoid leakage:** always tune on the **training set only** (that's what the internal CV does). The test set stays untouched until the final comparison; otherwise the reported score is over-optimistic.

> **Why `model__alpha`?** The double underscore means "set the `alpha` parameter *inside* the step named `model`." That is how a grid search reaches into a Pipeline.

<font color='black'><div class="alert alert-success"><strong>**Tune alpha with cross-validation on TRAIN data only**</strong><br>

````python
ridge_search = GridSearchCV(
    Pipeline([
        ("scaler", RobustScaler()),
        ("model", Ridge(random_state=RANDOM_STATE)),
    ]),
    param_grid={"model__alpha": [0.01, 0.1, 1.0, 10.0, 100.0]},
    scoring="neg_root_mean_squared_error",
    cv=5,
    n_jobs=-1,
    #verbose=3,# 0 (default) Silent, 1 Shows number of fits + total time, 2 Shows each fold/candidate as it runs, 3 Adds scores and timing per fit
)
ridge_search.fit(X_train, y_train)
print("Best alpha:", ridge_search.best_params_["model__alpha"])
print(f"CV RMSE: {-ridge_search.best_score_}")

ridge_best = ridge_search.best_estimator_
y_pred_ridge = ridge_best.predict(X_test)
ridge_metrics = evaluate_regression(y_test, y_pred_ridge, "Ridge (tuned)")
````

<font color='black'><div class="alert alert-success"><strong>**Random Forest Regressor**</strong><br>

<font color='black'><div class="alert alert-warning"><strong>**Concept: Random Forest Regressor**</strong><br>

**What is it?**

A **Random Forest** builds **many decision trees** and averages their predictions.

**One decision tree** asks yes/no questions:
- *Is MedInc > 3.5?* → go left
- *Is Latitude < 35?* → go right
- … until it reaches a leaf with a price prediction

**Random Forest** A Random Forest builds many decision trees — each on a random sample of rows and features — and averages their predictions to reduce overfitting and variance.

**Intuition:** Like asking 200 junior analysts for a price estimate and taking the median — individual trees overfit, but the **ensemble** (combination of several models) is stable.

| Pros | Cons |
|------|------|
| Handles non-linear patterns and interactions | Harder to explain than linear models |
| Needs little feature scaling (we still scale for fair comparison) | Slower on huge data |
| Robust default choice | Can overfit if trees are too deep |

**Key hyperparameters:** `n_estimators` (how many trees), `max_depth` (how deep each tree grows).

<font color='black'><div class="alert alert-success"><strong>**RandomForestRegressor Model Pipeline**</strong><br>

````python
rf_pipe = Pipeline([
    ("scaler", RobustScaler()),
    ("model", RandomForestRegressor(
        n_estimators=200,
        max_depth=12,
        min_samples_leaf=5,
        random_state=RANDOM_STATE,
        n_jobs=-1,
        #verbose=2,
    )),
])

rf_pipe.fit(X_train, y_train)
y_pred_rf = rf_pipe.predict(X_test)
````

<font color='black'><div class="alert alert-success"><strong>**Model Evaluation:- RandomForestRegressor**</strong><br>

````python
rf_metrics = evaluate_regression(y_test, y_pred_rf, "Random Forest")
plot_predictions(y_test, y_pred_rf, "Random Forest: Actual vs Predicted (test set)")
````

<font color='black'><div class="alert alert-success"><strong>**Feature importance — how much each feature reduced impurity across trees**</strong><br>

````python
rf_model = rf_pipe.named_steps["model"]
importance = pd.Series(rf_model.feature_importances_, index=FEATURE_NAMES).sort_values(ascending=False)

plt.figure(figsize=(8, 4))
importance.plot(kind="barh", color="forestgreen")
plt.title("Random Forest feature importance")
plt.xlabel("Importance")
plt.tight_layout()
plt.show()
````

<font color='black'><div class="alert alert-success"><strong>**Tune n_estimators and max_depth with cross-validation on TRAIN data only**</strong><br>

````python
rf_search = GridSearchCV(
    Pipeline([
        ("scaler", RobustScaler()),
        ("model", RandomForestRegressor(random_state=RANDOM_STATE, n_jobs=-1)),
    ]),
    param_grid={
        "model__n_estimators": [10,20, 70, 100, 200],
        "model__max_depth": [8, 12, None],
    },
    scoring="neg_root_mean_squared_error",
    cv=5,
    n_jobs=-1,
)
rf_search.fit(X_train, y_train)
print("Best params:", rf_search.best_params_)

rf_best = rf_search.best_estimator_
y_pred_rf = rf_best.predict(X_test)
rf_metrics = evaluate_regression(y_test, y_pred_rf, "Random Forest (tuned)")
````

<font color='black'><div class="alert alert-success"><strong>**Gradient Boosting Regressor**</strong><br>

<font color='black'><div class="alert alert-warning"><strong>**Concept: Gradient Boosting Regressor**</strong><br>

**What is it?**

**Gradient Boosting** like RandomForestRegressor, also uses trees, but **sequentially** — each new tree tries to fix the **errors** of the previous ones.

1. Start with a simple prediction (e.g. mean price).
2. Build tree #1 to predict **residuals** (actual − prediction).
3. Add tree #1's output (scaled by learning rate) to the prediction.
4. Repeat for hundreds of trees.

**Intuition:** A student takes a practice exam (tree 1), sees wrong answers, studies those topics (tree 2), repeats until scores plateau. This often beats Random Forest on structured tabular data.

| Pros | Cons |
|------|------|
| Often best accuracy on tabular data | More hyperparameters to tune |
| Captures complex patterns | Easier to overfit if trees too deep / lr too high |
| Industry workhorse (XGBoost, LightGBM cousins) | Slower to train than Ridge |

**Key hyperparameters:** `n_estimators`, `learning_rate` (η), `max_depth`.

> **Note:** In production many teams use **XGBoost** or **LightGBM** — same idea, optimized implementation. We use sklearn's `GradientBoostingRegressor` so you need zero extra installs.

<font color='black'><div class="alert alert-success"><strong>**Gradient Boosting Regressor Model Pipeline**</strong><br>

````python
gb_pipe = Pipeline([
    ("scaler", RobustScaler()),
    ("model", GradientBoostingRegressor(
        n_estimators=200,
        learning_rate=0.1,
        max_depth=4,
        subsample=0.8,
        random_state=RANDOM_STATE,
    )),
])
gb_pipe.fit(X_train, y_train)
y_pred_gb = gb_pipe.predict(X_test)
````

<font color='black'><div class="alert alert-success"><strong>**Model Evaluation:- Gradient Boosting Regressor**</strong><br>

````python
gb_metrics = evaluate_regression(y_test, y_pred_gb, "Gradient Boosting")
plot_predictions(y_test, y_pred_gb, "Gradient Boosting: Actual vs Predicted (test set)")
````

<font color='black'><div class="alert alert-success"><strong>**Feature importance**</strong><br>

````python
gb_model = gb_pipe.named_steps["model"]
gb_importance = pd.Series(gb_model.feature_importances_, index=FEATURE_NAMES).sort_values(ascending=False)

plt.figure(figsize=(8, 4))
gb_importance.plot(kind="barh", color="darkorange")
plt.title("Gradient Boosting feature importance")
plt.xlabel("Importance")
plt.tight_layout()
plt.show()
````

<font color='black'><div class="alert alert-success"><strong>**Tune n_estimators and max_depth with cross-validation on TRAIN data only**</strong><br>

````python
gb_search = GridSearchCV(
    Pipeline([
        ("scaler", RobustScaler()),
        ("model", GradientBoostingRegressor(random_state=RANDOM_STATE)),
    ]),
    param_grid={
        "model__n_estimators": [50, 60, 70, 100, 200],
        "model__learning_rate": [0.05, 0.08, 0.1, 0.2],
        "model__max_depth": [2,3, 4, 5],
    },
    scoring="neg_root_mean_squared_error",
    cv=3,
    n_jobs=-1,
)
gb_search.fit(X_train, y_train)
print("Best params:", gb_search.best_params_)

gb_best = gb_search.best_estimator_
y_pred_gb = gb_best.predict(X_test)
gb_metrics = evaluate_regression(y_test, y_pred_gb, "Gradient Boosting (tuned)")
````

<font color='black'><div class="alert alert-warning"><strong>**Compare all three models**</strong><br>


Let's compare on the **same held-out test set** — the only fair comparison.

````python
results = pd.DataFrame([ridge_metrics, rf_metrics, gb_metrics])
results = results.sort_values("RMSE")
results
````

````python
fig, ax = plt.subplots(1, 3, figsize=(12, 4))
metrics = ["RMSE", "MAE", "R2"]
for i, m in enumerate(metrics):
    ax[i].bar(results["model"], results[m], color=["#4C72B0", "#55A868", "#C44E52"])
    ax[i].set_title(m)
    ax[i].tick_params(axis="x", rotation=25)
plt.tight_layout()
plt.show()

best_name = results.iloc[0]["model"]
print(f"\nLowest test RMSE: {best_name}")
````

<font color='black'><div class="alert alert-warning"><strong>**Cross-validation (stability check)**</strong><br>

A single train/test split can be lucky or unlucky. **5-fold cross-validation** trains on 5 different splits and averages RMSE.

We do this on **training data only** — the test set stays untouched until final sign-off.

````python
models = {
    "Ridge": ridge_best,
    "Random Forest": rf_best,
    "Gradient Boosting": gb_best,
}

cv_rows = []
for name, pipe in models.items():
    scores = cross_val_score(
        pipe, X_train, y_train,
        scoring="neg_root_mean_squared_error",
        cv=5,
        n_jobs=-1,
    )
    cv_rows.append({
        "Model": name,
        "CV RMSE mean": -scores.mean(),
        "CV RMSE std": scores.std(),
    })

cv_df = pd.DataFrame(cv_rows).sort_values("CV RMSE mean")
cv_df
````

<font color='black'><div class="alert alert-warning"><strong>**Residual analysis (did we miss a pattern?)**</strong><br>

**Residuals** = actual − predicted. Good models have residuals randomly scattered around zero — no curve, no funnel shape.

````python
best_pipe = gb_best if "Gradient" in best_name else (rf_best if "Forest" in best_name else ridge_best)
y_pred_best = best_pipe.predict(X_test)
residuals = y_test - y_pred_best

fig, axes = plt.subplots(1, 2, figsize=(11, 4))
axes[0].scatter(y_pred_best, residuals, alpha=0.3, s=10)
axes[0].axhline(0, color="red", linestyle="--")
axes[0].set_xlabel("Predicted")
axes[0].set_ylabel("Residual")
axes[0].set_title("Residuals vs Predicted")

axes[1].hist(residuals, bins=40, color="steelblue", edgecolor="white")
axes[1].set_title("Residual distribution")
plt.tight_layout()
plt.show()
````

**Which model and when best to use?**

| Situation | Start with |
|-----------|------------|
| Need interpretability / regulatory sign-off | **Ridge** (or Lasso) |
| Quick strong baseline, tabular data | **Random Forest** |
| Maximum accuracy on structured data | **Gradient Boosting** / XGBoost / LightGBM |
| Very large dataset, many features | **LightGBM** or **XGBoost** |
